# 01. Hyperspectral Image (HSI) Data Exploration
This notebook explores HSI data cubes, spectral signatures, spatial bands, and patch generation for QueenNet.

```python
# Install dependencies in Colab if needed
# !pip install pennylane torch numpy scipy scikit-learn matplotlib pandas seaborn pyyaml
import sys
sys.path.append('../src')

import numpy as np
import matplotlib.pyplot as plt
from queennet.data.loaders import generate_synthetic_hsi
from queennet.data.preprocessing import normalize_spectral
from queennet.data.splits import split_hsi_samples
```

```python
# Generate sample HSI Cube
image, gt = generate_synthetic_hsi(height=64, width=64, bands=103, num_classes=9, seed=42)
print(f'HSI Cube Shape: {image.shape} (Height, Width, Spectral Bands)')
print(f'Ground Truth Map Shape: {gt.shape}')

# Visualize false color composite & ground truth
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))
ax1.imshow(image[:, :, [50, 27, 17]] / image.max())
ax1.set_title('False Color Composite (Bands 50, 27, 17)')
ax1.axis('off')

im2 = ax2.imshow(gt, cmap='tab10')
ax2.set_title('Ground Truth Class Regions')
ax2.axis('off')
plt.colorbar(im2, ax=ax2, fraction=0.046, pad=0.04)
plt.tight_layout()
plt.show()
```
