# 02. QueenNet End-to-End Quantum Smoke Test
Google Colab & CPU-compatible verification of QueenNet's Quantum Encoding, Quantum Convolution, and Quantum Measurement.

```python
import sys
sys.path.append('../src')

import torch
import pennylane as qml
from queennet.models.queennet import QueenNet

print(f'PyTorch: {torch.__version__} | PennyLane: {qml.__version__}')
print(f'Device: {"cuda" if torch.cuda.is_available() else "cpu"}')
```

```python
# Test 12-qubit QueenNet forward and backward pass
model = QueenNet(in_bands=103, num_classes=2, num_qubits=12, depth_L=2, patch_size=32)
x = torch.randn(2, 103, 32, 32)
logits = model(x)
loss = logits.sum()
loss.backward()

print(f'Forward Logits Shape: {logits.shape}')
print(f'QC Weights Grad Norm: {model.qc.weights.grad.norm().item():.6f}')
print('Parameter Breakdown:', model.count_parameters())
```
